```python
import numpy as np
import matplotlib
from matplotlib import pyplot as plt
```

```python
Hf = 1000 # [m]
H = 0 # [m]
B = 150 # [year] decay constant in our model

T = 1000 # [year] maximum time we choose to run this function until
dt1 = 0.1 # [year] time step
t = np.arange(0,T,dt1) # [year]
```

The analytical solution is:
$$
H = H_f(1 - e^{(-\frac{t}{B})})
$$

```python
H_analytical = Hf * (1 - np.exp(-t / B) )
```

```python
fig = plt.figure(figsize=(6,4))
axes = fig.add_subplot(1,1,1)
axes.plot(t, H_analytical, 'r')
axes.set_xlabel("time (years)")
axes.set_ylabel("ice height (m)")
axes.set_title("Analytical Solution for Ice Height (H)")
```

The numeric solution is:
$$
H^{j+1} = \frac{dH}{dt} \Delta t + H^j
$$

```python
H_numerical = np.zeros_like(t) # make a zero array that has the same shape as t

for i in np.arange(0, len(t)-1):
    H_numerical[i+1] = H_numerical[i] + dt1 * (Hf - H_numerical[i]) / B
```

```python
fig = plt.figure(figsize=(6,4))
axes = fig.add_subplot(1,1,1)
axes.plot(t, H_numerical, 'b')
axes.set_xlabel("time (years)")
axes.set_ylabel("ice height (m)")
axes.set_title("Numerical Solution for Ice Height (H)")
```

Compute the difference between the analytical and the numerical models

```python
diff = H_analytical - H_numerical
RMS_diff = np.sqrt(np.mean(diff**2))
print(f"The root mean square error is: {RMS_diff}")
```
